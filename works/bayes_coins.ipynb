```python
# Task 3: Bayes' Theorem for 4 Coins - Probability After 9 Flips
def run_bayes_update(coins, initial_beliefs, tosses):
    # input: list of coin biases, initial beliefs, and observed tosses
    # output: final beliefs after processing all tosses, and list of next heads probabilities after each toss
    beliefs = initial_beliefs[:]
    print(f"Initial beliefs: {beliefs}")
    print("-" * 40)
    
    next_heads_probs = []

    for i, toss in enumerate(tosses):
        likelihoods = []
        for bias in coins:
            if toss == 1:
                likelihoods.append(bias)
            else:
                likelihoods.append(1 - bias)

        unnormalized = [beliefs[j] * likelihoods[j] for j in range(len(coins))]
        total = sum(unnormalized)
        beliefs = [val / total for val in unnormalized]

        next_heads = sum(beliefs[j] * coins[j] for j in range(len(coins)))
        next_heads_probs.append(round(next_heads, 2))

        # toss_name = "Heads" if toss == 1 else "Tails"
        # print(f"Toss {i+1} ({toss_name}): {[round(x, 4) for x in beliefs]}")
    
    return beliefs, next_heads_probs

coin_biases = [0.12, 0.27, 0.21, 0.96]
prior_beliefs = [0.25, 0.25, 0.25, 0.25]
observed_tosses = [1, 1, 1, 0, 1, 0, 1, 1, 1] # 1 for Heads, 0 for Tails

final_beliefs, next_heads_probs = run_bayes_update(coin_biases, prior_beliefs, observed_tosses)


print(f"Next H probabilities toss: {next_heads_probs}")
print("-" * 40)
print(f"Final probabilities (rounded): {[round(b, 2) for b in final_beliefs]}")
print("-" * 40)
print(f"Sum check: {sum(final_beliefs)}")
```
