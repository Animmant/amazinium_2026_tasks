```python
words = [
    'Листя', 'зелене', ',', 'птахи', 'співають', ',', 'а', 'все', 'навіть', 'квітне', 'навкруги', '.'
]

signs = [',', '.', '!', '?']
sentence = ''

for i in range(len(words)):
    sentence += words[i]
    # First, check if this is not the last element, and only then check the next one
    if i < len(words) - 1 and words[i+1] not in signs:
        sentence += ' '

print(sentence)
```
