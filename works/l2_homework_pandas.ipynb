```python
import pandas as pd
import numpy as np

file_sci = 'scimagojr.xlsx'
file_ei = 'Energy Indicators.xls'

def answer_one():
    ## skipping the first 17 rows and the last 38 rows, and using only columns C to F, and treating '...' as NaN
    energy = pd.read_excel(file_ei, skiprows=17, skipfooter=38, usecols='C:F', na_values='...')

    name_label_energy = {
        'Unnamed: 2': 'Country',
        'Petajoules': 'Energy Supply',
        'Gigajoules': 'Energy Supply per Capita',
        '%': '% Renewable'
    }
    name_country_energy = {
        'Republic of Korea': 'South Korea',
        'United States of America': 'United States',
        'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
        'China, Hong Kong Special Administrative Region': 'Hong Kong'
    }

    energy.rename(columns=name_label_energy, inplace=True)
    # Remove everything inside the parentheses "(), numbers", including the parentheses themselves
    energy['Country'] = energy['Country'].str.replace(r'\([^)]*\)|\d+', '', regex=True).str.strip()
    energy['Country'] = energy['Country'].replace(name_country_energy)

    # energy['Country'] = energy['Country'].str.replace(r'\d+|\s*\(.*\)', '', regex=True).str.strip()
    # energy = energy.rename(columns=name_label_energy)

    print(energy.head())
    return energy


answer_one()
    
```
