### Text classification of search query 

</font>

- **python**  as snake -> Zoology
- **python**  as programming language -> Computer Science
- **python**  as "monty python" -> Entertainment

Probabilistic model:

#### Bayes Rule

\begin{equation*}
P(y|X) = \frac{P(X| y) \cdot P(y)}{P(X)} 
\quad\quad\quad
Posterior = \frac{ Likelihood \cdot Prior}{Evidence} 
\quad\quad\quad
P(class| python) = \frac{P(python| class) \cdot P(class)}{P(python)} 
\end{equation*}

Considering the $P(python)$ is common for all classes we may compare just nominators: 

\begin{equation*}
P(python| Zoology) \cdot P(Zoology) 
\quad\quad\quad 
P(python|CS) \cdot P(CS) 
\quad\quad\quad
P(python|Entertainment) \cdot P(Entertainment) 
\end{equation*}

In general: 
\begin{equation*}
\hat{y} =  \underset{y}{argmax} \quad P(y|X) =  \underset{argmax}{y} P(X|y) \cdot P(y)
\end{equation*}

Most probably predicted class is <font color = blue>CS</font>

#### Naive Bayes Classifiers

\begin{equation*}
\hat{y} =  \underset{y}{argmax} \quad P(y) \prod_{ i=1 }^{ n }{ P(x_{ i }|\,y) } 
\end{equation*}

If search query = **"python snake"** 
\begin{equation*}
\hat{y} =  \underset{y}{argmax} \quad
P(y)\cdot P(python|\,y) \cdot P(snake|\,y)
\end{equation*}

Now, the most probably predicted class is <font color = blue>Zoology</font> since  $P(snake|\,CS)$ is far less than $P(snake|\,Zoology)$

Note: if one of word is not presented in text then its statistical propability = 0 and as the result
the whole likelihood = 0 regardless of other words. Thus it is worth using laplace smooting  


<font color = green >

#### Laplace smooting 
 
</font>


$
A : 1 \quad
B : 3\quad
C : 0\quad
D : 6\quad
$

$N= 10\quad K =4$ 
<br>N - number of samples, K - number of classes

\begin{equation*}
P(A) = 0.1\quad\quad\quad\quad
P(B) = 0.3\quad\quad\quad\quad
P(C) = 0.0\quad\quad\quad\quad
P(D) = 0.6\\
\end{equation*}

<font color = blue >

\begin{equation*}
P^{\,L}(x_{i}) =  \frac{P(x_{i})+1}{N+K}
\end{equation*}

</font>




\begin{equation*}
P^{\,L}(A) =  \frac{1+1}{10+4} = 0.14 \quad P^{\,L}(B) =  \frac{3+1}{10+4} = 0.29
\quad P^{\,L}(C) =  \frac{0+1}{10+4} = 0.07 \quad P^{\,L}(D) =  \frac{6+1}{10+4} = 0.5
\end{equation*}

## Term Frequency–Inverse Document Frequency (TF-IDF)

</font>

TF-IDF is a numerical statistic that reflects how important a word is to a document in a collection or corpus. Its value **increases** with the number of times a word appears in a document and **decreases** with the number of documents in the corpus that contain that word.

<div style="float:left;">
<br>

### Term Frequency

The term frequency \(tf(t,d)\) measures how often term \(t\) appears in document \(d\):

$$
tf(t,d) = \frac{k}{n}
$$  

\(d\) — document;  
\(k\) — number of times the word occurs in \(d\);  
\(n\) — total number of words in \(d\).

*Augmented frequency* (to reduce bias toward longer documents):

$$
tf^{\,A}(t,d) = 0.5 + 0.5 \frac{tf(t,d)}
                     {\max_{t' \in d} tf(t',d)}
$$

### Inverse Document Frequency

The inverse document frequency \(idf(t,D)\) measures how much information the word provides:

$$
idf(t,D) = \log \frac{N}{K}
$$  

\(D\) — the entire collection of documents;  
\(K\) — number of documents in \(D\) that contain the word;  
\(N\) — total number of documents in \(D\).

</div>

**Note:** Various approaches can be used for inverse document frequency.

For **“this”**:

$$
tf(\text{"this"}, d_{1}) = \frac{1}{5} = 0.2, \quad
tf(\text{"this"}, d_{2}) = \frac{1}{7} \approx 0.14, \quad
idf(\text{"this"}, D) = \log\frac{2}{2} = 0
$$

$$
tfidf(\text{"this"}, d_{1}, D) = 0.2 \times 0 = 0, \qquad
tfidf(\text{"this"}, d_{2}, D) = 0.14 \times 0 = 0
$$

For **“example”**:

$$
tf(\text{"example"}, d_{1}) = \frac{0}{5} = 0, \quad
tf(\text{"example"}, d_{2}) = \frac{3}{7} \approx 0.43, \quad
idf(\text{"example"}, D) = \log\frac{2}{1} \approx 0.30
$$

$$
tfidf(\text{"example"}, d_{1}, D) = 0 \times 0.30 = 0, \qquad
tfidf(\text{"example"}, d_{2}, D) = 0.43 \times 0.30 \approx 0.129
$$

### Key Takeaways

1. **Evolution of Sentiment Analysis**:

### 1. Data Preparation & Splitting

</font>

We merge the read lists `texts_neg` and `texts_pos`, assign them the appropriate labels (0 for negative, 1 for positive) and split them into `train` and `test`.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

labels_neg = [0] * len(texts_neg)
labels_pos = [1] * len(texts_pos)

texts = texts_neg + texts_pos
labels = labels_neg + labels_pos

df = pd.DataFrame({
    'text': texts,
     'label': labels
})

# stratify=df['label'] ensures the balance between classes is maintained
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], 
    df['label'], 
    test_size=0.3, 
    random_state=42, 
    stratify=df['label']
)

print(f"Train size: {len(X_train)}")
print(f"Test size: {len(X_test)}")

print("label distribution in training set:")
print(y_train.value_counts())
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from pandas import Series
import numpy as np

def evaluate_and_print_metrics(method_name: str, y_true: Series, y_pred: np.ndarray) -> None:
    """
    Calculates and prints standard classification metrics to avoid code duplication.
    """
    print(f"{'='*40}")
    print(f"--- {method_name} ---")
    print(f"Accuracy:  {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"Recall:    {recall_score(y_true, y_pred):.4f}")
    print(f"F1-score:  {f1_score(y_true, y_pred):.4f}")
    print(f"{'='*40}\n")
```

```python
evaluate_and_print_metrics("Baseline (Random Guessing)", y_test, np.random.choice([0, 1], size=len(y_test)))
```

### 3. Method 1: TF-IDF + Logistic Regression

</font>

Baseline method. We use `min_df=5` to filter out rare words (noise) that appear in fewer than 5 documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

tfidf_vect = TfidfVectorizer(
    min_df=5, 
    max_features=10000
)

X_train_tfidf = tfidf_vect.fit_transform(X_train)
X_test_tfidf = tfidf_vect.transform(X_test)

# Train baseline model
lr_tfidf = LogisticRegression(max_iter=1000, random_state=42)
lr_tfidf.fit(X_train_tfidf, y_train)

# Predict and evaluate
preds_tfidf = lr_tfidf.predict(X_test_tfidf)
evaluate_and_print_metrics("Method 1: TF-IDF + Logistic Regression", y_test, preds_tfidf)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

ngram_vect = CountVectorizer(
    ngram_range=(1, 2), 
    min_df=5, 
    max_features=20000
)

X_train_ngram = ngram_vect.fit_transform(X_train)
X_test_ngram = ngram_vect.transform(X_test)

lr_ngram = LogisticRegression(max_iter=1000, random_state=42)
lr_ngram.fit(X_train_ngram, y_train)

preds_ngram = lr_ngram.predict(X_test_ngram)
evaluate_and_print_metrics("Method 2: N-grams (1,2) + Logistic Regression", y_test, preds_ngram)
```

### 5. Method 3: Word Embeddings (Word2Vec) + Logistic Regression

</font>

3-ird. A modern approach based on dense vectors. The model generates its own embeddings by learning the semantic proximity of words directly from our training corpus.

```python
# import nltk
# from nltk.tokenize import word_tokenize

# from nltk.tokenize import sent_tokenize
# # Обов'язково завантажуємо моделі для токенізації
# nltk.download('punkt') 

# from nltk.tokenize import word_tokenize

# text = "I'm learning AI in 2026. It's awesome!"
# tokens = word_tokenize(text=text, preserve_line=True)

# print(tokens)
# # Результат: ['I', "'m", 'learning', 'AI', 'in', '2026', '.', 'It', "'s", 'awesome', '!']
```

```python
import nltk
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec

nltk.download('punkt', quiet=True)

# Tokenize texts for Word2Vec training
X_train_tok = [word_tokenize(text.lower(), preserve_line=True) for text in X_train] # preserve_line=True to keep punctuation as separate tokens
X_test_tok = [word_tokenize(text.lower(), preserve_line=True) for text in X_test]

# Train Word2Vec embeddings on our specific domain (movie reviews)
w2v_model = Word2Vec(
    sentences=X_train_tok, 
    vector_size=100, 
    window=5, 
    min_count=2, 
    workers=4
)

def get_document_vector(doc: list[str], model: Word2Vec) -> np.ndarray:
    """
    Averages word vectors to create a single document vector representation.
    """
    valid_words = [word for word in doc if word in model.wv]
    if not valid_words:
        return np.zeros(model.vector_size)
    return np.mean([model.wv[word] for word in valid_words], axis=0)

# Vectorize train and test sets using the trained Word2Vec model
X_train_w2v = np.array([get_document_vector(doc, w2v_model) for doc in X_train_tok])
X_test_w2v = np.array([get_document_vector(doc, w2v_model) for doc in X_test_tok])

# Train model on Word2Vec features
lr_w2v = LogisticRegression(max_iter=1000, random_state=42)
lr_w2v.fit(X_train_w2v, y_train)

# Predict and evaluate
preds_w2v = lr_w2v.predict(X_test_w2v)
evaluate_and_print_metrics("Method 3: Word2Vec + Logistic Regression", y_test, preds_w2v)
```
